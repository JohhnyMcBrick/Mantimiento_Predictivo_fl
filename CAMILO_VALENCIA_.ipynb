```python
import pandas as pd
from ydata_profiling import ProfileReport

df = pd.read_csv('ai4i2020.csv')
df.head()
```

```python
print("Filas y columnas:", df.shape)
print("\nTipos de datos:")
print(df.dtypes)
print("\nResumen estadístico:")
df.describe()
```

```python
print("Duplicados:", df.duplicated().sum())
print("\nValores nulos por columna:")
print(df.isnull().sum())
```

```python
profile = ProfileReport(df, title="Exploracion - Predictive Maintenance Dataset")
profile.to_file("reporte.html")

from google.colab import files
files.download("reporte.html")
```

Observaciones generales:

- El dataset contiene 10.000 registros y 14 columnas.
- No se encontraron valores nulos en ninguna columna.
- No se encontraron filas duplicadas.
- La variable objetivo (Machine failure) esta desbalanceada: la gran mayoria
  de registros corresponden a la clase 0 (sin falla), lo cual es esperado en
  contextos de mantenimiento real.
- Air temperature y Process temperature presentan correlacion alta entre si
  (0.864), lo que indica que ambas variables se comportan de manera similar.
- Torque y Rotational speed tienen correlacion negativa fuerte (-0.916),
  lo que significa que a mayor velocidad de rotacion, menor es el torque aplicado.
- HDF y Machine failure presentan correlacion moderada-alta (0.573), siendo
  HDF uno de los tipos de falla con mayor relacion con la falla general.
- Tool wear muestra correlacion moderada con Machine failure (0.213),
  lo que sugiere que el desgaste acumulado influye en la probabilidad de falla.
- Los tipos de datos son correctos: variables numericas continuas para
  sensores y binario para Machine failure.
